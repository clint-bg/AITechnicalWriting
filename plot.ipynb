```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('surveyresults1.csv',header=0)
```

```python
df
```

## 345 Question 2

```python
q2 = df[['Q2-SD','Q2-D','Q2-U','Q2-A','Q2-SA']].iloc[0]
#rename columns
q2 = q2.rename(index={
    'Q2-SD':'Strongly Disagree',
    'Q2-D':'Disagree',
    'Q2-U':'Undecided',
    'Q2-A':'Agree',
    'Q2-SA':'Strongly Agree'
})
```

```python
#plot bar chart
plt.figure(figsize=(10, 6))
plt.grid(alpha=0.3,zorder=1)
plt.bar(q2.index, q2.values*100, color='seagreen', zorder=2)
#plt.title('ChEn 345: It is beneficial to my learning to use large language models (LLMs) like ChatGPT, Gemini, Claude, or others to help me draft and improve my lab report.')
plt.xlabel('Response')
plt.ylabel('Percentage of Respondents (Total Count = 74)')
plt.ylim(0,70)
plt.show()
```
